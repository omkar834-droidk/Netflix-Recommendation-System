```python
df = df.dropna(subset = ['title'])
df  = df.dropna(subset = ['vote_average'])
```

```python

import ast

df['genres'] = df['genres'].apply(lambda x :" "   .join(i['name'] for i in ast.
literal_eval(x)))
```

```python
df['tagline'] = df['tagline'].fillna('')
```

```python
df['tags'] = df['overview'] + df['tagline'] + df['genres']
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text):
    text = str(text.lower())
    text = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    words = text.split()
    words = [word for word in words if word not in stop_words]

    words = [lemmatizer.lemmatize(word) for word in words]

    return ' '.join(words)
```

```python
df['tags'][0]
```

```python
df['tags'] = df['tags'].apply(preprocess_text)
```

```python
df['tags'][0]
```

```python
df  = df.reset_index(drop = True)
```

```python
movie_indices = pd.Series(df.index, index = df['title']).drop_duplicates()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=5000 , ngram_range=(1,2))
stop_words = ('english')
```

```python
tidf_matrix = tfidf.fit_transform(df['tags'])
```

```python
tidf_matrix.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title , n = 10 ):
  if title not in movie_indices:
    return ["Movie not found in the dataset."]
  idx = movie_indices[title]
  similarity_scores = cosine_similarity(tidf_matrix[idx] , tidf_matrix).flatten()
  similar_movies_indices = similarity_scores.argsort()[::-1][1:n+1]
  return df['title'].iloc[similar_movies_indices]

recommend('Toy Story' , 10)
```

```python
import pickle

# Save the preprocessed DataFrame
with open('df.pkl', 'wb') as f:
    pickle.dump(df, f)
print("Saved df.pkl")

# Save the movie_indices Series
with open('movie_indices.pkl', 'wb') as f:
    pickle.dump(movie_indices, f)
print("Saved movie_indices.pkl")

# Save the TfidfVectorizer object
with open('tfidf.pkl', 'wb') as f:
    pickle.dump(tfidf, f)
print("Saved tfidf.pkl")

# Save the TF-IDF matrix
# Note: tidf_matrix is a sparse matrix, so we can save it directly
with open('tidf_matrix.pkl', 'wb') as f:
    pickle.dump(tidf_matrix, f)
print("Saved tidf_matrix.pkl")
```
